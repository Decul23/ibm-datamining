# Kursus 1: Exploratory Data Analysis (EDA) for Machine Learning
Modul ini mencakup *Exploratory Data Analysis* (EDA) secara menyeluruh: pembersihan data, penanganan nilai hilang, analisis statistik, rekayasa fitur (*feature engineering*), dan visualisasi data.

## 1. Import Library & Pembuatan Dataset Simulasi
Kami akan membuat dataset sintetis perilaku pelanggan *e-commerce*.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
np.random.seed(42)

# Membuat dataset eksplorasi e-commerce
n_samples = 500
data = {
    'CustomerID': range(1001, 1001 + n_samples),
    'Age': np.random.randint(18, 65, size=n_samples),
    'AnnualIncome': np.random.normal(55000, 15000, size=n_samples),
    'SpendingScore': np.random.randint(1, 100, size=n_samples),
    'PurchaseFrequency': np.random.poisson(lam=5, size=n_samples),
    'CategoryPreference': np.random.choice(['Electronics', 'Fashion', 'Home', 'Beauty'], size=n_samples)
}

df = pd.DataFrame(data)
# Menambahkan missing values secara acak untuk simulasi pembersihan data
df.loc[df.sample(frac=0.05).index, 'AnnualIncome'] = np.nan
df.head()
```

## 2. Pemeriksaan Struktur & Pembersihan Data Missing

```python
print("--- Informasi Dataset ---")
print(df.info())
print("\n--- Jumlah Missing Value ---")
print(df.isnull().sum())

# Imputasi missing value menggunakan nilai median
income_median = df['AnnualIncome'].median()
df['AnnualIncome'].fillna(income_median, inplace=True)
print("\nMissing value setelah imputasi:", df['AnnualIncome'].isnull().sum())
```

## 3. Visualisasi Analisis Eksploratif (EDA) & Korelasi

```python
plt.figure(figsize=(14, 4))

plt.subplot(1, 3, 1)
sns.histplot(df['AnnualIncome'], kde=True, color='skyblue')
plt.title('Distribusi Pendapatan Tahunan')

plt.subplot(1, 3, 2)
sns.boxplot(x='CategoryPreference', y='SpendingScore', data=df, palette='Set2')
plt.title('Skor Pengeluaran per Kategori')

plt.subplot(1, 3, 3)
sns.scatterplot(x='AnnualIncome', y='SpendingScore', hue='CategoryPreference', data=df)
plt.title('Pendapatan vs Skor Pengeluaran')

plt.tight_layout()
plt.show()

# Matrix Korelasi Numerik
plt.figure(figsize=(6, 4))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='Blues')
plt.title('Matriks Korelasi Fitur')
plt.show()
```
